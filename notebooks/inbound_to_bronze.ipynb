```scala
val anuncio_dados = spark.read.format("json").load("/Volumes/databricks_curso/default/inbound/dados_brutos_imoveis.json")
display(anuncio_dados)
```

```scala
val dados_anuncio = anuncio_dados.drop("imagens", "usuario")
display(dados_anuncio)
```

```scala
import org.apache.spark.sql.functions.col
```

```scala
val df_bronze = dados_anuncio.withColumn("id", col("anuncio.id"))
display(df_bronze)
```

```scala
df_bronze.write
  .format("delta")
  .mode("append")
  .saveAsTable("databricks_curso.bronze.imoveis_raw")
```
